1. Tampilkan citra hanya pada kombinasi kanal Merah-Biru dan kanal HijauBiru saja, dengan kanal yang tidak dipakai dinolkan. Susun keduanya berdampingan dalam satu figure yang diberi judul.

```python
# Read citra dan konversi dari BGR ke RGB
img = cv.imread('ktm.jpg')
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# 1. Kombinasi Merah-Biru (Kanal Hijau / Index 1 dinolkan)
img_rb = img_rgb.copy()
img_rb[:, :, 1] = 0

# 2. Kombinasi Hijau-Biru (Kanal Merah / Index 0 dinolkan)
img_gb = img_rgb.copy()
img_gb[:, :, 0] = 0

# Subplot 1: Merah-Biru
plt.subplot(1, 2, 1)
plt.imshow(img_rb)
plt.title('Kanal Merah-Biru (Hijau = 0)')
plt.axis('off')

# Subplot 2: Hijau-Biru
plt.subplot(1, 2, 2)
plt.imshow(img_gb)
plt.title('Kanal Hijau-Biru (Merah = 0)')
plt.axis('off')
# Menampilkan keduanya berdampingan dalam satu figure
plt.figure(figsize=(12, 6))

plt.suptitle('Perbandingan Kombinasi Kanal Warna', fontsize=16)
plt.show()
```

2. Tampilkan citra dalam tiga bentuk pencerminan: vertikal, horizontal, dan
keduanya, dalam satu figure berisi tiga subplot yang masing-masing diberi
label.

```python
img_rgb = cv.cvtColor(cv.imread('ktm.jpg'), cv.COLOR_BGR2RGB)

# Flip: 0 (vertikal), 1 (horizontal), -1 (keduanya)
img_v, img_h, img_b = cv.flip(img_rgb, 0), cv.flip(img_rgb, 1), cv.flip(img_rgb, -1)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
titles = ['Vertikal', 'Horizontal', 'Keduanya']
images = [img_v, img_h, img_b]

for ax, img_res, title in zip(axes, images, titles):
    ax.imshow(img_res)
    ax.set_title(title)
    ax.axis('off')

plt.show()
```

**Studi Kasus 1** : Panitia sebuah kegiatan perlu membubuhkan label identitas pada
ratusan foto dokumentasi. Foto berasal dari beberapa kamera dan ponsel,
sehingga ukurannya bermacam-macam. Label harus terlihat seimbang pada
semua foto tanpa disetel satu per satu. Fungsi yang sama dikenakan pada dua
citra berbeda ukuran. Tinggi bar dan ukuran huruf ikut menyesuaikan, sehingga
proporsinya tetap sama. Silakan melakukan akuisisi data sendiri, setiap anak
harus menggunakan image yang berbeda-beda

```python
def beri_label_adaptif(img_path):
    img = cv.imread(img_path)
    h, w = img.shape[:2]

    bar_h = int(h * 0.08)
    font_scale = h / 500

    cv.rectangle(img, (0, h - bar_h), (w, h), (0, 0, 0), -1)
    cv.putText(img, 'DOKUMENTASI ACARA', (20, h - int(bar_h * 0.3)),
               cv.FONT_HERSHEY_SIMPLEX, font_scale, (255, 255, 255), 2)
    return cv.cvtColor(img, cv.COLOR_BGR2RGB)

plt.imshow(beri_label_adaptif('ktm.jpg'))
plt.axis('off')
plt.show()
```

**Studi Kasus 2** : Marketplace seperti Tokopedia dan Shopee menampilkan foto
produk dalam bingkai persegi 1:1. Foto penjual yang berukuran potrait atau
lanskap akan terpotong atau terlihat tidak proporsional bila diunggah begitu saja.
Bangun program yang mengubah foto berukuran apa pun menjadi thumbnail
persegi tanpa mengubah rasio aslinya, dengan sisa ruang diisi kanvas berwarna.
Dalam hal ini, rasio asli tetap terjaga, citra terpasang tepat di tengah kanvas,
tambahkan logo nama toko berukuran proporsional pada gambar.

```python
def make_square_thumbnail(img_path, shop_name="TOKO SAYA"):
    img = cv.imread(img_path)
    h, w = img.shape[:2]
    size = max(h, w)

    # Kanvas putih persegi
    canvas = np.full((size, size, 3), 255, dtype=np.uint8)

    # Posisi tengah
    y_off, x_off = (size - h) // 2, (size - w) // 2
    canvas[y_off:y_off+h, x_off:x_off+w] = img

    # nama toko di pojok kiri atas
    cv.putText(canvas, shop_name, (int(size*0.03), int(size*0.07)),
               cv.FONT_HERSHEY_SIMPLEX, size/600, (0, 0, 0), 2)

    return cv.cvtColor(canvas, cv.COLOR_BGR2RGB)

plt.imshow(make_square_thumbnail('ktm.jpg'))
plt.axis('off')
plt.show()
```
